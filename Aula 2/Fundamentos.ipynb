### 2. Estruturas de Dados (Listas, Tuplas e Dicionários)
Essas são as estruturas mais utilizadas para organizar informações.

```python
# Listas (Mtáveis)
frutas = ["maça", "banana", "cereja"]
frutas.append("uva")

# Tupla (Inutável)
coordenadas = (10, "Vini")

# Dicionário (Chave-valor)
aluno = {"nome": "Tiago", "curso": "Tecnologia", "nota": 9.5}

print(f"Lista: {frutas}")
print(f"\n")
print(f"Dicionário: {aluno['nome']} tirou {aluno['nota']}")
```

#### 3. Estruturas de controle e repetição

controle de fluxo de execução do programa

```python
# condicional
media = 7.5
if media >= 6:
    resultado = "Aprovado"
else: resultado = "Reprovado"

# loop for
numeros = [1, 2, 3, 4, 5]
quadrados = [x**2 for x in numeros]

print(f"Resultado : {resultado}")
print(f"Quadrados : {quadrados}")

print("\nLista de Números (jeito longo)")
quadrados = []
for x in numeros:
    quadrados.append(x**2)
print(quadrados)
```

## 4. Funções e Lambda

Modularização de código e funções anônimas

```python
def calcular_bonus(salario, percentual = 10):
    return salario + (salario * percentual / 100)

# função Lambda
dobro = lambda x: x * 2

print(f"Bônus: {calcular_bonus(3000)}")
print(f"Dobro de 5: {dobro(5)}")
```
