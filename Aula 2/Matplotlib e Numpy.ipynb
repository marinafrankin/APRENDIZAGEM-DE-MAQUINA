```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# A partir de uma lista
meu_array = np.array([1, 2, 3, 4, 5])
print("Meu array:", meu_array)

#Criando um array de zeros
zeros = np.zeros((2, 3)) # Matriz 2x33 de zeros
print("Array de zeros:\n", zeros)

# Criando um array com um intervalo
intervalo = np.arange(0, 11, 2) # De 0 a 10, de 2 em 2
print("Array com intervalo:\n", intervalo)
```

```python
array_a = np.array([1, 2, 3])
array_b = np.array([4, 5, 6])

soma = array_a + array_b
multiplicacao = array_a * array_b

print("Soma: ", soma)
print("Multiplicação: ", multiplicacao)
```

```python
anos = [2020, 2021, 2022, 2023, 2024]
vendas = [100, 120, 150, 130, 160]

plt.plot(anos, vendas)
plt.title('Vendas ao Longo dos Anos')
plt.xlabel('Ano')
plt.ylabel('Vendas (em milhões)')
plt.grid(True)
plt.show()
```

```python
produtos = ['Produto A', 'Produto B', 'Produto C']
quantidades = [50, 80, 65]

plt.bar(produtos, quantidades)
plt.title('Quantidade de Produtos em Estoque')
plt.xlabel('Produto')
plt.ylabel('Quantidade')
plt.show()
```

```python
np.random.seed(42)
x = np.random.rand(50) * 10
y = 2 * x + np.random.randn(50) * 2

plt.scatter(x, y)
plt.title('Relação entre X  e Y')
plt.xlabel('Variável X')
plt.ylabel('Variável Y')
plt.show()
```

```python
# Gerando dados com Numby
x = np.linspace(0, 2 * np.pi, 100)
y_seno = np.sin(x)
y_cosseno = np.cos(x)
#y_cosseno2 = np.cos(x*0.2)

# Criando o gráfico com Matplotlib
plt.plot(x, y_seno, label='Seno')
plt.plot(x, y_cosseno, label='Cosseno')
#plt.plot(x, y_cosseno2, label='Cosseno2')
plt.title('Funções Seno e Cosseno')
plt.xlabel('Ângulo [rad]')
plt.ylabel('Vaor')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Gera 1000 números aleatórios com média (loc) 0 e desvio padrão (scale) 1
dados = np.random.normal(loc=0, scale=1, size=1000)

# Cria o histograma
# 'bins' define o número de barras no gráfico
plt.hist(dados, bins=30, color='skyblue', edgecolor='black')

# Adiciona títulos e rótulos
plt.title('Histograma de uma Distribuição Normal')
plt.xlabel('Valores')
plt.ylabel('Frequência')

# Exibe o gráfico
plt.show()
```

```python
# Cria um array de 100 pontos de -2*PI a 2*PI
x = np.linspace(-2 * np.pi, 2 * np.pi, 200)

# Cria uma figura e uma grade de subplots 2x2
# 'fig' é a figura inteira, 'axs' é um array 2x2 com os eixos de cada gráficos
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Adiciona um título principal para todos os subplots
fig.suptitle('Múltiplos Gráficos (Subplots)', fontsize=16)

# Gráfico 1 (Canto superior esquerdo)
axs[0, 0].plot(x, np.sin(x), color='blue')
axs[0, 0].set_title('Seno')
axs[0, 0].grid(True)

# gráfico 2 (canto superior direito)
axs[0, 1].plot(x, np.cos(x), color='red')
axs[0, 1].set_title('Cosseno')
axs[0, 1].grid(True)

# Gráfico 3 (canto inferior esquerdo)
axs[1, 0].plot(x, np.tan(x), color='green')
axs[1, 0].set_title('Tangente')
axs[1, 0].set_ylim(-5, 5) # Limita o eixo y para melhor visualização
axs[1, 0].grid(True)

# Gráfico 4 (canto inferior direito)
axs[1, 1].plot(x, x**2, color='purple')
axs[1, 1].set_title('Parábola (x^2)')
axs[1, 1].grid(True)

# Ajusta o layout para evitar a sobreposição de títulos
plt.tight_layout(rect=[0, 0, 1, 0.96])

# Exibe a figura com todos os gráficos
plt.show()
```

```python
# Dados categóricos
categorias = ['Tecnologia', 'Saúde', 'Educação', 'Transporte']
valores = [45, 25, 15, 15]

explode = (0.1, 0, 0, 0)

plt.pie(valores, explode=explode, labels=categorias, autopct='%1.1f%%', shadow=True, startangle=90)

plt.axis('equal')

plt.title('Distribuição de Investimentos por Setor')
plt.show()
```
