### 2º Passo: Preparando o Dataset

Nosso dataset está em um arquivo CSV chamado dados.csv, contendo três colunas:

- idade
- salario
- compra (1 para compra e 0 para não compra)

```python
# Função para carregar dados do CSV
def carregar_dados(arquivo):
    X, y = [], []
    with open(arquivo, 'r') as f:
        leitor = csv.reader(f)
        next(leitor) #pular cabeçalho
        for linha in leitor:
            X.append([float(linha[0]), float(linha[1])]) #idade, salario
            y.append(int(linha[2]))
        return np.array(X), np.array(y)

# Carregando o dataset
X, y = carregar_dados("dataset.csv")
print("Formato dos dados:", X.shape, y.shape)
print("Primeiras amostras:\n", X[:5], "\nRótulos:", y[:5])
```

### 3º Passo: Medida de Impureza -> Entropia

```python
def entropia(y):
    classes, contagem = np.unique(y, return_counts=True)
    probs = contagem / len(y)
    return -np.sum(probs * np.log2(probs + 1e-9))

# Teste da entropia
print("Entropia [0,0,0,0] ->", entropia([0,0,0,0]))
print("Entropia [0,1,0,1] ->", entropia([0,1,0,1]))
```

### 4º Passo: Divisão dos Dados

```python
def dividir(X, y, features, limiar):
    esquerda_idx = X[:, features] <= limiar
    direita_idx = X[:, features] > limiar
    return X[esquerda_idx], y[esquerda_idx], X[direita_idx], y[direita_idx]

# Testando divisao por idade <= 40
Xe, ye, Xd, yd = dividir(X, y, features=0, limiar=40)
print("Esquerda:", ye)
print("Direita:", yd)
```
